# K Means Clustering


## Import the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Import the dataset

```python
dataset = pd.read_csv('/content/Mall_Customers.csv')
X = dataset.iloc[:,[3,4]].values # choosing the two features to find patters in
```

## Elbow Method
> to find the optimal number of clusters(k value)

### Calculating the WCSS(within cluster sum of squares) values for each k

```python
from sklearn.cluster import KMeans
wcss = []
for i in range(1,11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)
```

```python
print(wcss)
```

### Plotting WCSS vs k

```python
# X axis
k = list(range(1,11))
print(k)
```

```python
plt.plot(k,wcss)
plt.title('The Elbow Plot')
plt.xlabel('No. of Clusters')
plt.ylabel('WCSS')
```

Form the plot we can see the **optimal number of clusters** is **5**

## Train the K means model on the dataset

```python
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42)
y_kmeans = kmeans.fit_predict(X)
```

```python
print(y_kmeans)
```

## Visualising the clusters

```python
plt.scatter(X[y_kmeans==0, 0], X[y_kmeans==0, 1], s=50, c='red', label='Cluster 1')
plt.scatter(X[y_kmeans==1, 0], X[y_kmeans==1, 1], s=50, c='blue', label='Cluster 2')
plt.scatter(X[y_kmeans==2, 0], X[y_kmeans==2, 1], s=50, c='green', label='Cluster 3')
plt.scatter(X[y_kmeans==3, 0], X[y_kmeans==3, 1], s=50, c='yellow', label='Cluster 4')
plt.scatter(X[y_kmeans==4, 0], X[y_kmeans==4, 1], s=50, c='brown', label='Cluster 5')
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1], s=100, c='cyan', label='Centroids')

plt.title('K Means Clusters')
plt.xlabel('Annual Income(k$)')
plt.ylabel('Spending Score(1-100)')
plt.legend()
plt.show()
```
